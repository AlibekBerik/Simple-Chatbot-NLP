```python
!pip install nltk

import nltk
nltk.download('punkt')
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import re


qa_pairs = {

    "hello": "Hi there! How can I help you?",
    "hi": "Hello! Nice to meet you!",
    "hey": "Hey! How's it going?",
    "good morning": "Good morning! How are you today?",
    "good afternoon": "Good afternoon! How are you?",

    "what is your name": "I am ChatBot, nice to meet you!",
    "who are you": "I am a chatbot created using machine learning.",
    "how old are you": "I am timeless! I don't age like humans.",
    "where are you from": "I exist in the cloud, everywhere and nowhere.",

    "what can you do": "I can answer questions, tell jokes, and chat with you.",
    "can you tell me a joke": "Why did the computer go to the doctor? Because it caught a virus!",
    "can you help me": "Of course! I will try my best to help you.",
    "give me advice": "Always keep learning, even if it is just a little every day.",

    "how are you": "I am doing well, thank you!",
    "are you okay": "Yes! I am fine and ready to chat.",
    "what's your mood": "I am always happy to talk with you!",
    "tell me something interesting": "Did you know that honey never spoils?",
    "what's your favorite color": "I like all colors equally.",
    "do you like music": "Yes! Music is beautiful.",
    "do you have friends": "I have many friends who talk to me every day!",

    "what is the capital of france": "The capital of France is Paris.",
    "who discovered gravity": "Isaac Newton discovered gravity when an apple fell on his head.",
    "what is ai": "AI stands for Artificial Intelligence. It's when machines can learn and make decisions.",
    "what is machine learning": "Machine learning is a way for computers to learn from data without being explicitly programmed.",

    "bye": "Goodbye! See you again.",
    "see you later": "Have a nice day!",
    "talk to you later": "Farewell!"
}

questions = list(qa_pairs.keys())
answers = list(qa_pairs.values())


def preprocess(text):
    text = text.lower()
    text = re.sub(r"[^\w\s]", "", text)
    text = text.replace("u", "you")
    text = text.replace("whatsup", "what is up")
    text = text.strip()
    return text


vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(questions)


def get_response(user_input):
    user_input = preprocess(user_input)
    tfidf_user = vectorizer.transform([user_input])
    similarity = cosine_similarity(tfidf_user, tfidf_matrix)

    index = similarity.argmax()
    score = similarity[0][index]

    if score < 0.05:
        return "Sorry, I didn't understand that."

    return answers[index]


print("Chatbot: Hello! Type 'quit' to stop.")

while True:
    user = input("You: ")
    if user.lower() == "quit":
        print("Chatbot: Goodbye!")
        break
    print("Chatbot:", get_response(user))
```
